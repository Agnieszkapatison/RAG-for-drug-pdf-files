```python
import os
import fitz  # PyMuPDF
from langchain.text_splitter import RecursiveCharacterTextSplitter

# Ścieżka do folderu z plikami PDF
PDF_FOLDER = "C:\\Users\\Agnieszka\\OneDrive\\Pulpit\\RAG\\RAG-for-drug-pdf-files\\database"

# Zmienna do przechowywania tekstu ze wszystkich plików PDF
all_text = ""

# Iterowanie po wszystkich plikach PDF w folderze
for filename in os.listdir(PDF_FOLDER):
    if filename.endswith(".pdf"):
        pdf_path = os.path.join(PDF_FOLDER, filename)
        print(f"Przetwarzanie pliku: {filename}")
        
        try:
            # Otwieranie PDF za pomocą PyMuPDF (fitz)
            doc = fitz.open(pdf_path)
            
            # Iterowanie po stronach i zbieranie tekstu
            for page_num in range(doc.page_count):
                page = doc.load_page(page_num)
                all_text += page.get_text()  # Pobieranie tekstu z każdej strony
                
        except Exception as e:
            print(f"Błąd podczas przetwarzania pliku {filename}: {e}")

# Po zebraniu całego tekstu, dzielimy go na mniejsze fragmenty (chunki)
splitter = RecursiveCharacterTextSplitter(chunk_size=1500, chunk_overlap=100)
chunks = splitter.split_text(all_text)

# Drukowanie liczby fragmentów (chunks)
print(f"Liczba chunków: {len(chunks)}")

# Wyświetlenie pierwszego chunku
if chunks:
    print(f"Treść pierwszego chunku:\n{chunks[0]}")
```

```python
from langchain.embeddings.openai import OpenAIEmbeddings
from langchain.vectorstores import FAISS
import os

# Ładowanie klucza API z pliku .env
OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")
MODEL = os.getenv("MODEL")

# Inicjalizacja embeddingów za pomocą GPT (OpenAI)
embeddings = OpenAIEmbeddings(openai_api_key=OPENAI_API_KEY)

# Ścieżka do folderu OneDrive, gdzie będziesz zapisywać indeksy FAISS
ONEDRIVE_PATH = "C:\\Users\\Agnieszka\\OneDrive\\faiss_index"  # Zaktualizuj ścieżkę do swojego OneDrive

# Tworzenie vectorstore za pomocą FAISS z przetworzonych chunków
vectorstore = FAISS.from_texts(chunks, embeddings)

# Zapis indeksu FAISS w folderze OneDrive
vectorstore.save_local(ONEDRIVE_PATH)
```

```python

# Jeśli będziesz chciała później odczytać indeks FAISS z OneDrive:
# vectorstore = FAISS.load_local(ONEDRIVE_PATH, embeddings)

# Sprawdzenie działania na zapytaniu
query = "Jakie są działania niepożądane leku Ifapidin?"
docs = vectorstore.similarity_search(query, k=1)

# Wyświetlenie wyników
for i, doc in enumerate(docs):
    print(f"Result {i+1}: {doc.page_content}")
```

```python
retriever = vectorstore.as_retriever()
retriever.invoke("Jakie są działania niepożądane leku Ifapidin")
     
```

```python
from langchain.chat_models import ChatOpenAI
import os

# Ładowanie klucza API z pliku .env
OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")
MODEL = os.getenv("MODEL")  # np. 'gpt-3.5-turbo'

# Inicjalizacja modelu GPT z OpenAI
model = ChatOpenAI(openai_api_key=OPENAI_API_KEY, model=MODEL, temperature=0)

# Wykonanie zapytania za pomocą modelu GPT
response = model.invoke("Who is the president of the United States?")

# Wyświetlenie odpowiedzi modelu
print(response)
```

```python
from langchain_core.output_parsers import StrOutputParser

parser = StrOutputParser()

chain = model | parser 
print(chain.invoke("Who is the president of the United States?"))
```

```python
from langchain.prompts import PromptTemplate

template = """
You are an assistant that provides answers to questions based on
a given context. 

Answer the question based on the context. If you can't answer the
question, reply "I don't know".

Be as concise as possible and go straight to the point.

Context: {context}

Question: {question}
"""

prompt = PromptTemplate.from_template(template)
print(prompt.format(context="Here is some context", question="Here is a question"))
```

```python
chain = prompt | model | parser

chain.invoke({
    "context": "Anna's sister is Susan", 
    "question": "Who is Susan's sister?"
})

     
```

```python
from operator import itemgetter

chain = (
    {
        "context": itemgetter("question") | retriever,
        "question": itemgetter("question"),
    }
    | prompt
    | model
    | parser
)
```

```python
questions = [
    "Jak dawkuje się Ifapidin",
    "Jaką postać farmaceutyczną ma Arthrotec?",
    "Czy Bisocard obniża ciśnienie?",
]

for question in questions:
    print(f"Question: {question}")
    print(f"Answer: {chain.invoke({'question': question})}")
    print("*************************\n")
     
```
